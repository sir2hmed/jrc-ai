# Classification Metrics

Classification metrics measure how well a model predicts categories such as:

- Spam or not spam
- Fraud or legitimate
- Pass or fail
- Normal or suspicious

## Important Terms

Before understanding the metrics, we need to understand four possible prediction results:

| Term | Meaning |
|---|---|
| **True Positive (TP)** | The model correctly predicted the positive class |
| **True Negative (TN)** | The model correctly predicted the negative class |
| **False Positive (FP)** | The model predicted positive, but the actual class was negative |
| **False Negative (FN)** | The model predicted negative, but the actual class was positive |

For example, in suspicious-passenger detection:

- **TP:** A suspicious passenger was correctly detected.
- **TN:** A normal passenger was correctly classified as normal.
- **FP:** A normal passenger was incorrectly flagged as suspicious.
- **FN:** A suspicious passenger was incorrectly classified as normal.

---

## 1. Accuracy

Accuracy measures the percentage of all predictions that were correct.

### Equation

$$
\text{Accuracy} =
\frac{TP + TN}{TP + TN + FP + FN}
$$

### Example

If the model made 100 predictions and 90 were correct:

$$
\text{Accuracy} = \frac{90}{100} = 0.90 = 90\%
$$

### Python Code

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, predictions)
print("Accuracy:", accuracy)
```

### When to Use It

Accuracy is useful when the classes are balanced.

It can be misleading with unbalanced data. For example, if 95% of passengers are normal, a model that predicts everyone as normal will have 95% accuracy but will detect no suspicious passengers.

---

## 2. Precision

Precision measures how often the model is correct when it predicts the positive class.

It answers:

> Out of everything predicted as positive, how many were actually positive?

### Equation

$$
\text{Precision} = \frac{TP}{TP + FP}
$$

### Example

The model flagged 20 passengers as suspicious:

- 15 were actually suspicious.
- 5 were normal.

$$
\text{Precision} = \frac{15}{15 + 5} = 0.75 = 75\%
$$

### Python Code

```python
from sklearn.metrics import precision_score

precision = precision_score(y_test, predictions)
print("Precision:", precision)
```

### When to Use It

Precision is important when false positives or false alarms are costly.

---

## 3. Recall

Recall measures how many of the actual positive cases the model successfully detected.

It answers:

> Out of all actual positive cases, how many did the model find?

### Equation

$$
\text{Recall} = \frac{TP}{TP + FN}
$$

### Example

There were 20 suspicious passengers:

- The model detected 15.
- The model missed 5.

$$
\text{Recall} = \frac{15}{15 + 5} = 0.75 = 75\%
$$

### Python Code

```python
from sklearn.metrics import recall_score

recall = recall_score(y_test, predictions)
print("Recall:", recall)
```

### When to Use It

Recall is important when missing a positive case is dangerous or expensive.

Examples include:

- Detecting security threats
- Detecting diseases
- Detecting fraudulent transactions

---

## 4. F1-Score

The F1-score creates a balance between Precision and Recall.

### Equation

$$
\text{F1-score} =
2 \times
\frac{\text{Precision} \times \text{Recall}}
{\text{Precision} + \text{Recall}}
$$

### Example

If:

- Precision = 0.80
- Recall = 0.60

Then:

$$
\text{F1-score}
=
2 \times \frac{0.80 \times 0.60}{0.80 + 0.60}
=
0.686
$$

### Python Code

```python
from sklearn.metrics import f1_score

f1 = f1_score(y_test, predictions)
print("F1-score:", f1)
```

### When to Use It

F1-score is useful when:

- The dataset is unbalanced.
- Both false positives and false negatives are important.
- You want one score that combines Precision and Recall.

---

## 5. ROC-AUC

ROC-AUC measures how well the model separates the positive and negative classes across different classification thresholds.

**ROC** means Receiver Operating Characteristic.

**AUC** means Area Under the Curve.

### Interpretation

| ROC-AUC | General interpretation |
|---:|---|
| `0.50` | Similar to random guessing |
| `0.60–0.70` | Weak |
| `0.70–0.80` | Acceptable |
| `0.80–0.90` | Good |
| `0.90–1.00` | Excellent |

A higher ROC-AUC score means the model is generally better at ranking positive cases above negative cases.

### Python Code

```python
from sklearn.metrics import roc_auc_score

probabilities = model.predict_proba(X_test)[:, 1]
roc_auc = roc_auc_score(y_test, probabilities)

print("ROC-AUC:", roc_auc)
```

ROC-AUC should normally use probability or confidence scores rather than final class predictions.

---

## Confusion Matrix

A confusion matrix displays the number of correct and incorrect predictions.

|  | Predicted Negative | Predicted Positive |
|---|---:|---:|
| **Actual Negative** | True Negative (TN) | False Positive (FP) |
| **Actual Positive** | False Negative (FN) | True Positive (TP) |

### Python Code

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

ConfusionMatrixDisplay.from_predictions(y_test, predictions)
plt.show()
```

The confusion matrix helps identify whether the model produces too many false alarms or misses too many positive cases.

---

## Classification Report

A classification report displays the following metrics for every class:

- Precision
- Recall
- F1-score
- Support

**Support** is the number of actual examples belonging to each class.

### Python Code

```python
from sklearn.metrics import classification_report

report = classification_report(y_test, predictions)
print(report)
```

### Example Output

```text
              precision    recall  f1-score   support

Normal           0.90      0.95      0.92       100
Suspicious       0.75      0.60      0.67        20

accuracy                              0.89       120
```

| Metric | Simplest Term |
|---|---|
| **Precision** | Can I trust a positive prediction? |
| **Recall** | How many real cases did the model find? |
| **F1-score** | Are Precision and Recall balanced? |
| **Support** | How many examples were tested? |
| **Accuracy** | How many total predictions were correct? |
| **Macro average** | Treat every class equally. |
| **Weighted average** | Give larger classes more influence. |


| Metric | Question it answers |
|---|---|
| **Accuracy** | How many total predictions were correct? |
| **Precision** | When the model predicted positive, how often was it correct? |
| **Recall** | How many actual positive cases did the model detect? |
| **F1-score** | How well does the model balance Precision and Recall? |
| **ROC-AUC** | How well does the model separate positive and negative cases? |

# Regression Metrics

Regression metrics measure how well a model predicts numerical values such as:

- House prices
- Exam scores
- Sales amounts
- Temperatures
- Travel times

The main regression metrics are:

1. Mean Absolute Error (MAE)
2. Mean Squared Error (MSE)
3. R² Score

---

## Example

Suppose a model predicts three students' exam scores:

| Student | Actual score | Predicted score | Error |
|---|---:|---:|---:|
| Student 1 | 70 | 68 | 2 |
| Student 2 | 80 | 85 | -5 |
| Student 3 | 90 | 87 | 3 |

---

## 1. Mean Absolute Error (MAE)

MAE calculates the average absolute difference between the actual values and predicted values.

It answers:

> On average, how far are the predictions from the actual values?

### Equation

$$
\text{MAE} =
\frac{1}{n}
\sum_{i=1}^{n}
|y_i-\hat{y}_i|
$$

Where:

- $y_i$ = actual value
- $\hat{y}_i$ = predicted value
- $n$ = number of predictions
- $| \ |$ = absolute value, which removes negative signs

### Example

The absolute errors are:

$$
|70-68|=2
$$

$$
|80-85|=5
$$

$$
|90-87|=3
$$

Therefore:

$$
\text{MAE} =
\frac{2+5+3}{3}
=
3.33
$$

The model's predictions are approximately **3.33 marks away from the actual scores on average**.

### Python Code

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, predictions)
print("MAE:", mae)
```

### Interpretation

- A lower MAE is better.
- `MAE = 0` means perfect predictions.
- MAE uses the same unit as the predicted value.

For example, if the model predicts prices in Saudi riyals, MAE is also measured in Saudi riyals.

---

## 2. Mean Squared Error (MSE)

MSE calculates the average of the squared prediction errors.

It answers:

> How large are the model's errors when large mistakes receive an extra penalty?

### Equation

$$
\text{MSE} =
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
$$

### Example

First, square every error:

$$
(70-68)^2=4
$$

$$
(80-85)^2=25
$$

$$
(90-87)^2=9
$$

Then calculate the average:

$$
\text{MSE} =
\frac{4+25+9}{3}
=
12.67
$$

### Python Code

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, predictions)
print("MSE:", mse)
```

### Interpretation

- A lower MSE is better.
- `MSE = 0` means perfect predictions.
- Large errors receive a much stronger penalty because the errors are squared.
- MSE is sensitive to outliers.

MSE uses squared units. For example, if the model predicts marks, MSE is measured in squared marks, so it is less directly understandable than MAE.

---

## 3. R² Score

R² measures how much of the variation in the target values is explained by the model.

It answers:

> How well does the model explain the differences in the actual values?

### Equation

$$
R^2 =
1-
\frac{
\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
}{
\sum_{i=1}^{n}(y_i-\bar{y})^2
}
$$

Where:

- $y_i$ = actual value
- $\hat{y}_i$ = predicted value
- $\bar{y}$ = average of the actual values

### Example

For the exam-score example:

$$
R^2 = 1-\frac{38}{200}=0.81
$$

This means the model explains approximately **81% of the variation in the exam scores**.

### Python Code

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, predictions)
print("R²:", r2)
```

### Interpretation

| R² value | Meaning |
|---:|---|
| `1.00` | Perfect predictions |
| `0.80` | The model explains approximately 80% of the variation |
| `0.00` | The model is no better than predicting the average |
| Below `0` | The model is worse than predicting the average |

A higher R² score is generally better.

---

## Calculate All Regression Metrics

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

predictions = model.predict(X_test)

mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
r2 = r2_score(y_test, predictions)

print("MAE:", mae)
print("MSE:", mse)
print("R²:", r2)
```

---

## Quick Comparison

| Metric | Easiest way to remember it | Better result |
|---|---|---|
| **MAE** | How far away are the predictions on average? | Lower |
| **MSE** | How large are the errors after heavily penalizing big mistakes? | Lower |
| **R²** | How much of the target's variation does the model explain? | Higher |

---

## When to Use Each Metric

| Situation | Recommended metric |
|---|---|
| You want an error that is easy to understand | MAE |
| Large prediction mistakes should receive a stronger penalty | MSE |
| You want to measure how well the model explains the data | R² |
| You want a complete evaluation | Use MAE, MSE, and R² together |

## Simple Summary

- **MAE:** Average size of the model's mistakes.
- **MSE:** Average squared mistake, with a larger penalty for big errors.
- **R²:** How well the model explains the variation in the actual values.
